### Q40

a = [[]] * 3 -- It will create a list with three nested empty lists

a[1].append(5) -  5 will be appended to nested list at 2nd index postion

print(a) -- [[], [5], []]

I'm wrong 

* The * operator does not create three independent empty lists. Instead, it creates three   references pointing to the exact same single empty list in memory.
* Here is what goes on behind the scenes:
1. [[]] creates a list containing one empty list container.
2. * 3 copies the reference to that one empty list two more times.
3. Therefore, a[0], a[1], and a[2] are all just different names for the same identical list object.
When you run a[1].append(5), you are modifying that single shared list, so the change shows up in all three positions.

```python
a = [[]] * 3
a[1].append(5)
print(a)
```
