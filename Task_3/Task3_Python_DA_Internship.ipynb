### Q1. Find the Frequency of Each Word in a Sentence

Write a Python program to count how many times each word appears in a given sentence and store the result in a dictionary.

Example Input:

python is easy and python is powerful

Expected Output:

{'python': 2, 'is': 2, 'easy': 1, 'and': 1, 'powerful': 1}

* Step - 1 : Observe and Read The Problem
* Step - 2 : Understand the problem
* Step - 3 : whats happening to input to get the desired output
  * Input - string, Output - Dictionary of words and their frequency counts
* Step - 4 : Breakdown to simple steps
* Step - 5 : Logic - Pseudocode Intially
  * Convert the string to lowercase and split it into a list of words
  * Create an empty dictionary to store each word and its frequency
  * Create an empty set to keep track of words already encountered
  * Loop through each word in the list of words
  * Condition to check if the word is already present in the seen set:
    * If yes, increment the word's count in the dictionary by 1
    * If no, set the word's count to 1 in the dictionary and add the word to the seen set
  * Return the dictionary of word frequencies
* Step - 6 : Code
* Step - 7 : Test & debug and make changes in code till desired result is returned

```python
def freq_word(string : str):

    word_lst = string.lower().split() # Preparing a list of words from a string

    word_freq = {} # To store word and frequency

    seen = set() # To make note of already seen word to count the occurences
    
    for word in word_lst: # Loop to catch and use every word to count its occurences

        if word in seen: # Condition which checks if word is already seen 

            word_freq[word] += 1 # increase count if word is already seen

        else:

            word_freq[word] = 1 

            seen.add(word) # Adding word to seen when its first appeared

    return word_freq # Returning the word and its occurences
```

```python
freq_word('python is easy and python is powerful')
```

```python
freq_word('innomatics and innomatics')
```

### Q2. Find the Frequency of Each Character in a Word
Write a Python program to count how many times each character appears in a given word and store the result in a dictionary.

Example Input:
banana

Expected Output:
{'b': 1, 'a': 3, 'n': 2}

* Step - 1 : Observe and Read The Problem
* Step - 2 : Understand the problem
* Step - 3 : whats happening to input to get the desired output
  * Input - string word, Output - Dictionary of characters and their frequency counts (excluding spaces)
* Step - 4 : Breakdown to simple steps
* Step - 5 : Logic - Pseudocode Intially
  * Remove all spaces from the input string
  * Create an empty dictionary to store each character and its frequency
  * Create an empty set to keep track of characters already encountered
  * Loop through each character in the string
  * Condition to check if the character is already present in the seen set:
    * If yes, increment the character's count in the dictionary by 1
    * If no, add the character to the seen set and set its count to 1 in the dictionary
  * Return the dictionary of character frequencies
* Step - 6 : Code
* Step - 7 : Test & debug and make changes in code till desired result is returned

```python
def freq_char(word : str):

    word = word.replace(" ", "") # Remove spaces to count only character frequencies
    
    frequency = {} # Dictionary to store each character and its count

    seen = set() # Set to keep track of characters already encountered

    for char in word: # Loop through each character in the string

        if char in seen: # Check if the character has already been seen

            frequency[char] += 1 # Increment count if character already exists

        else:

            seen.add(char) # Mark character as seen for the first time

            frequency[char] = 1 # Initialize its count to 1

    return frequency # Return dictionary of character frequencies
```

### Q3. Find the First Non-Repeating Character in a Word
Write a Python program to find the first character that appears only once in a given word. If no such character exists, return None.

Example Input:

swiss

Expected Output:

w

* Step - 1 : Observe and Read The Problem
* Step - 2 : Understand the problem
* Step - 3 : whats happening to input to get the desired output
  * Input - string, Output - First non-repeating character, or None if all characters repeat
* Step - 4 : Breakdown to simple steps
* Step - 5 : Logic - Pseudocode Intially
  * Strip extra spaces from the start and end of the string
  * Create an empty dictionary to store character counts
  * Loop through each character in the string:
    * Condition to check if character is already in char_freq:
      * If yes, increment its count by 1
      * If no, set its count to 1
  * Loop through each key-value pair char, freq in char_freq:
    * Condition to check if freq == 1:
      * If yes, return char immediately
  * Return None if no character has a frequency of 1
* Step - 6 : Code
* Step - 7 : Test & debug and make changes in code till desired result is returned

```python
def first_non_repeat_char(word : str):

    word = word.strip() # Remove leading and trailing whitespaces

    char_freq = {} # Dictionary to store each character and its frequency

    for char in word: # First loop to count occurrences of each character

        if char in char_freq: # Check if character has already been recorded

            char_freq[char] += 1 # Increment frequency count if seen before

        else:

            char_freq[char] = 1 # Initialize frequency count to 1 for first appearance
            
    for char, freq in char_freq.items(): # Second loop to find the first unique character

        if freq == 1: # Condition to check if frequency is exactly one

            return char # Return the first character with a count of 1

    return None # Return None if all characters repeat
```

### Q6. Move All Zeros to the End of a List
Write a Python program to move all zeros to the end of a list while maintaining the original order of non-zero elements.

Example Input:

[0, 3, 0, 1, 2, 0, 4]

Expected Output:

[3, 1, 2, 4, 0, 0, 0]


* Step - 1 : Observe and Read The Problem
* Step - 2 : Understand the problem
* Step - 3 : whats happening to input to get the desired output
  * Input - list of numbers, Output - same list with all zeroes moved to the end
* Step - 4 : Breakdown to simple steps
* Step - 5 : Logic - Pseudocode Intially
  * Loop through the list using index from 0 to end of list
  * Condition to check if number at current index is zero:
    * If yes, pop that element from its index
    * Append that popped zero at the end of the list
  * Return the modified list
* Step - 6 : Code
* Step - 7 : Test & debug and make changes in code till desired result is returned

```python
def zeroes_at_end(nums : list):

    for i in range(len(nums)): # Loop through the list using index

        if nums[i] == 0: # Check if current number is zero

            x = nums.pop(i) # Remove zero from its current position

            nums.append(x) # Add the removed zero at the end of the list

    return nums # Return modified list with zeroes at end
```

### Q7. Find the Second Highest Distinct Number in a List

Write a Python program to find the second-largest distinct number in a given list.
Duplicate values should not be considered separate ranks.

If a second distinct number does not exist, return None.

Example Input:
[10, 20, 40, 30, 40, 20]

Expected Output:
30

* Step - 1 : Observe and Read The Problem
* Step - 2 : Understand the problem
* Step - 3 : whats happening to input to get the desired output
  * Input - list of numbers, Output - second highest unique number from list (or None)
* Step - 4 : Breakdown to simple steps
* Step - 5 : Logic - Pseudocode Intially
  * Create two variables largest and second_largest and set both to negative infinity
  * Loop through each number in nums list:
    * Condition 1: check if current number is greater than largest:
      * Put current largest value into second_largest
      * Put current number into largest
    * Condition 2: else check if current number is greater than second_largest and not equal to largest:
      * Update second_largest with current number
  * Check if second_largest is still negative infinity:
    * If yes, return None
  * Return second_largest
* Step - 6 : Code
* Step - 7 : Test & debug and make changes in code till desired result is returned

```python
def second_highest(nums : list):

    largest = float('-inf') # Variable to keep track of largest number

    second_largest = float('-inf') # Variable to keep track of second largest number

    for num in nums: # Loop through every number in the list

        if num > largest: # If current number is bigger than largest so far

            second_largest = largest # Old largest becomes second largest
            
            largest = num # Update largest with new number

        elif num > second_largest and num != largest: # If number is smaller than largest but greater than second largest

            second_largest = num # Update second largest

    if second_largest == float('-inf'): # If second largest never got updated (all numbers same or list too small)

        return None

    
    return second_largest # Return the second largest number found
```

```python
second_highest([100, 200, 50, 400])
```

### Q8. Remove Duplicates Without Changing the Original Order
Write a Python program to remove duplicate elements from a list while preserving the order of their first occurrence.

Example Input:

[4, 2, 4, 5, 2, 7, 5]

Expected Output:

[4, 2, 5, 7]

* Step - 1 : Observe and Read The Problem
* Step - 2 : Understand the problem
* Step - 3 : whats happening to input to get the desired output
  * Input - list of numbers, Output - list without duplicates in same order
* Step - 4 : Breakdown to simple steps
* Step - 5 : Logic - Pseudocode Intially
  * need a set seen to remember visited numbers
  * need a new list result to store unique numbers in order
  * loop to iterate through each number in nums:
    * condition to check if number is not in seen:
      * add number to seen
      * append number to result
  * return result list
* Step - 6 : Code
* Step - 7 : Test & debug and make changes in code till desired result is returned

```python
def remove_dups_same_order(nums : list):

    seen = set() # to track already visited numbers

    result = [] # list to store unique numbers in same order

    for num in nums: # loop to catch and check every number

        if num not in seen: # check if number is appearing first time

            seen.add(num) # add to seen so we know its visited

            result.append(num) # store in result list

    return result # return list with no duplicates in same order
```

### Q9. Find the Missing Number from 1 to N

Write a Python program to find the missing number in a list containing numbers from 1 to N.
The list contains exactly one missing number, and all remaining numbers appear only once.

Example Input:

nums = [1, 2, 4, 5, 6]

N = 6

Expected Output:

3

* Step - 1 : Observe and Read The Problem
* Step - 2 : Understand the problem
* Step - 3 : whats happening to input to get the desired output
  * Input - list of numbers from 1 to n with one number missing, Output - the missing number
* Step - 4 : Breakdown to simple steps
* Step - 5 : Logic - Pseudocode Intially
  * find actual_sum by summing all elements in nums list
  * find n by taking len(nums) + 1 because one number is missing from total count
  * calculate expected_sum using formula n * (n + 1) // 2
  * subtract actual_sum from expected_sum to get missing_number
  * return missing_number
* Step - 6 : Code
* Step - 7 : Test & debug and make changes in code till desired result is returned

```python
def missing_number(nums : list):

    actual_sum = sum(nums) # total sum of numbers present in the list

    n = len(nums) + 1 # actual count of numbers including the missing one
    
    expected_sum = n * (n + 1) // 2 # formula to get sum of first n natural numbers

    missing_number = expected_sum - actual_sum # difference gives the missing number

    return missing_number # returning the missing number
```

```python
missing_number([1, 2, 4, 5, 6])
```

```python
missing_number([1, 3, 4, 5, 6])
```

### Q10. Find Two Numbers Whose Sum Equals a Target

Write a Python program to find the first pair of numbers whose sum equals a given target.
Return the pair as a list. If no valid pair exists, return an empty list.

Return the pair whose second element is encountered earliest when scanning the input from left to right. If multiple earlier elements qualify, choose the earliest one.

Example Input:

nums = [2, 7, 11, 15]
target = 9

Expected Output:
[2, 7]

* Step - 1 : Observe and Read The Problem
* Step - 2 : Understand the problem
* Step - 3 : whats happening to input to get the desired output
  * Input - list of numbers and a target integer, Output - first pair as a list whose second element appears earliest or empty list
* Step - 4 : Breakdown to simple steps
* Step - 5 : Logic - Pseudocode Intially
  * create empty set seen to remember visited numbers
  * loop through each number in nums from left to right:
    * calculate complement = target - num
    * condition to check if complement is in seen:
      * return complement, num as a list
    * add current number to seen set
  * if loop ends without finding any pair, return empty list []
* Step - 6 : Code
* Step - 7 : Test & debug and make changes in code till desired result is returned

```python
def two_sum_pair(nums : list, target : int):

    seen = set() # set to store numbers we have seen already

    for num in nums:

        complement = target - num # needed number to reach target

        if complement in seen: # check if needed first element was already seen earlier

            return [complement, num] # returning pair of numbers as a list

        seen.add(num)

    return [] # returning empty list if no pair makes the target
```

```python
two_sum_pair([1, 2, 4, 5, 6], 6)
```

Write a Python program to find the number closest to zero in a given list.
If two numbers are equally close to zero, return the positive number.

Example Input:

[-8, -3, 2, 5, -2]

Expected Output:

2

* Step - 1 : Observe and Read The Problem
* Step - 2 : Understand the problem
* Step - 3 : whats happening to input to get the desired output
  * Input - list of numbers, Output - number closest to zero (positive number wins tie)
* Step - 4 : Breakdown to simple steps
* Step - 5 : Logic - Pseudocode Intially
  * initialize closest with first number
  * loop through each num in nums:
    * condition 1: if abs(num) is strictly less than abs(closest):
      * update closest with num
    * condition 2: else if abs(num) equals abs(closest) and num is greater than closest:
      * update closest with num
  * return closest
* Step - 6 : Code
* Step - 7 : Test & debug and make changes in code till desired result is returned

```python
def num_closest_to_zero(nums : list):

    closest = nums[0] # assuming the first number is the closest to zero

    for num in nums: # loop through every number in the list

        if abs(num) < abs(closest): # if present number closest to 0

            closest = num


        elif abs(num) == abs(closest) and num > closest: # For picking positive num if both nums are equally close

            closest = num

    return closest # return the closest number
```

### Q13. Rotate a List to the Right by K Positions
Write a Python program to rotate the elements of a list to the right by K positions.
Elements shifted beyond the last position should move to the beginning of the list.

Example Input:

nums = [1, 2, 3, 4, 5]

K = 2

Expected Output:

[4, 5, 1, 2, 3]

* Step - 1 : Observe and Read The Problem
* Step - 2 : Understand the problem
* Step - 3 : whats happening to input to get the desired output
  * Input - list of numbers and an integer k, Output - list rotated to the right by k positions
* Step - 4 : Breakdown to simple steps
* Step - 5 : Logic - Pseudocode Intially
  * check if nums is empty, if yes return nums
  * find length n using len(nums)
  * update k = k % n to avoid extra cycles
  * slice last k elements using nums[-k:]
  * slice remaining starting elements using nums[:-k]
  * join both slices and return
* Step - 6 : Code
* Step - 7 : Test & debug and make changes in code till desired result is returned

```python
def rotate_list(nums : list, k : int):

    if not nums: # if list is empty return as it is

        return nums

    n = len(nums) # length of the list

    k = k % n # adjust k if it is greater than length of list

    return nums[-k:] + nums[:-k] # take last k elements and attach remaining elements after them
```

```python
rotate_list([1, 2, 3, 4, 5], 2)
```

```python
rotate_list([4, 5, 1, 2, 3], 2)
```

### Q14. Find the Maximum Sum of Two Consecutive Elements

Write a Python program to find the maximum sum of any two consecutive elements in a list.
Assume the list contains at least two elements.

Example Input:

[4, 7, 2, 9, 5]

Expected Output:

14

* Step - 1 : Observe and Read The Problem
* Step - 2 : Understand the problem
* Step - 3 : whats happening to input to get the desired output
  * Input - list of numbers, Output - maximum sum of two consecutive numbers (or None if list has less than 2 elements)
* Step - 4 : Breakdown to simple steps
* Step - 5 : Logic - Pseudocode Intially
  * condition to check if len(nums) is less than 2:
    * return None
  * initialize highest_sum with sum of first two elements (nums[0] + nums[1])
  * loop i from index 0 to len(nums) - 2:
    * calculate current_sum = nums[i] + nums[i + 1]
    * condition to check if current_sum is greater than highest_sum:
      * update highest_sum with current_sum
  * return highest_sum
* Step - 6 : Code
* Step - 7 : Test & debug and make changes in code till desired result is returned

```python
def max_sum_consecutive(nums : list):

    if len(nums) < 2: # Checking if minimum two nums available in list

        return None

    highest_sum = nums[0] + nums[1] # Intializing sum with first and second num in list 

    for i in range(len(nums)-1):

        current_sum = nums[i] + nums[i+1] # sum of two adjacent elements

        if current_sum > highest_sum: # if current sum is bigger, update highest sum

            highest_sum = current_sum

    return current_sum # return the max consecutive sum
```

### Q15. Find All Pairs of Numbers Whose Sum Equals a Target


Write a Python program to find all unique pairs of numbers whose sum equals a given target.
Each pair should contain two distinct list positions, and duplicate pairs should not be included.
Return each pair in ascending numerical order. Return the list of pairs sorted by their first element. If no valid pairs exist, return an empty list.

Example Input:

nums = [2, 4, 3, 5, 7, 8, 1]

target = 9

Expected Output:

[[1, 8], [2, 7], [4, 5]]


```python
def pairs_equal_target(nums : list, target : int):

    unique_pairs = set() # set to avoid duplicate pairs

    for i in range(len(nums)): 

        for j in range(i + 1, len(nums)):

            if nums[i] + nums[j] == target: # checking if sum matches target

                # make sure pair is in ascending order
                pair = (min(nums[i], nums[j]), max(nums[i], nums[j]))

                unique_pairs.add(pair)

    
    result = [list(p) for p in sorted(unique_pairs)] # converting tuples back to lists and sort

    return result
```

```python
pairs_equal_target( [2, 4, 3, 5, 7, 8, 1], 9)
```

```python
pairs_equal_target( [2, 4, 3, 5, 7, 8, 1], 10)
```
