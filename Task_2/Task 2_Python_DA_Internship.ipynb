### Q1. Find the Shortest Word in a Sentence
Write a Python program to find the shortest word in a given sentence.

Example input:
Python makes coding easy

Expected output:
easy

* Step - 1 : Observe and Read The Problem
* Step - 2 : Understand the problem
* Step - 3 : whats happening to input to get the desired output
 * Input - string, Output - Smallest word - Single Word 
* Step - 4 :  Breakdown to simple steps
* Step - 5 : Logic - Pseudocode Intially
    * We need a varaible to store and update the smallest_word
    * loop to iterate along the each and every word in a sentence
    * breakdown and store the string as individual words in a list
    * Condition to check and update the smallest word
    * return the smallest word
* Step - 6: Code
* Step - 7 : Test & debug and make changes in code till desired result is returned

```python
def shortest_word(text : str): # text is a parameter that stores sentence which is a string

    text_list = text.split() # splits the text string into individual words and stores them as a list

    shortest_word = text_list[0] # assuming the first word in the list of words as shortest first to compare 
                                 # with others
    for i in text_list: # loop for checking each and every word against other
        if len(i) < len(shortest_word): # condition thats checking if current_word length is less than previous word
            shortest_word = i # updating the smallest word
    
    return shortest_word # returning the shortest word
                
```

Write a Python program to find the first longest word in a sentence. If multiple words have the same maximum length, return the one that appears first.

 * Input - string, Output - First_Longest_word - single word

* Step - 5 : Logic - Pseudocode Intially
    * A variable to split the text and store the individual words as a list
    * We need a varaible to store the lengths of each and every word

    * Condition to check and return the longest word

* Step - 6: Code
* Step - 7 : Test & debug and make changes in code till desired result is returned


```python
def first_long_word(text : str): # accpets the string as an input
    words_list = text.split() # saving words in text as individual words in a list 
    list_lens = [] # to store lengths of each word to compare and return the first longest word

    for i in words_list: # 1st loop for adding lengths of each word to a list of lengths
        
        list_lens.append(len(i)) # adding individual lengths to a list
        
    for j in words_list: # 2nd loop for comparing each word length with highest length
        
        if len(j) == max(list_lens): # comparing each word length with highest length of all words
            
            return j # Returning the first longest word
```

### Q6. Count Vowels in a String
Write a Python program to count the total number of vowels in a string. Consider both uppercase and lowercase vowels.

Example input:

Expected output:
5

* Step - 1 : Observe and Read The Problem
* Step - 2 : Understand the problem
* Step - 3 : whats happening to input to get the desired output
 * Input - string, Output - Count_Vowels - Single Number
* Step - 4 :  Breakdown to simple steps
* Step - 5 : Logic - Pseudocode Intially
    * A variable to store vowels in a string for comparing each and every char from string if its vowel or not
    * A varaible named count to count number of vowels
    * loop to iterate along the each and every char in a string to compare if its vowel or not
    * Condition to check each and every char if its vowel or not
    * if yes increase the count by 1
    * return the count of vowels
* Step - 7 : Test & debug and make changes in code till desired result is returned

```python
def count_vowels(text : str): # function that counts vowels where text takes user input string

    vowels = "aeiou" # predefined string with vowels in it to check against text chars if they are vowels

    count = 0 # Intiating as 0 if any vowel is encountered increase it by one to count no of vowels

    for char in text: # loop to traverse through each char in text 
        
        if char.lower() in vowels: # Condition that checks if its vowel or not
            
            count += 1 # if char in text count of vowel increased by 1
            
    return count # Finally returning how many chars in text are vowels
```

    * A variable to store vowels in a string for comparing each and every char from string if its vowel or not because chars other than vowels are consonants

    * if yes its a consonant increase the count by 1
    * return the count of consonants
* Step - 7 : Test & debug and make changes in code till desired result is returned

```python
def count_consonants(text : str): # function that counts consonants where text takes user input string

    vowels = "aeiou" # predefined string with vowels in it to check against text chars if they are vowels
                     # If they are not vowels they are consonants
    
    count_consonant = 0 # Initializing as 0 if its a consonant count increases
    
    for char in text: # loop to traverse through each char in text 
        
        if char.lower() not in vowels and char.isalpha(): # Condition that checks if its alphabet and not vowel i.e its consonant 
                                                  # sapces, digits and special chars ignored by isalpha()
            count_consonant += 1 # if char not in vowels count of consonant increased by 1
            
    return count_consonant # Finally returning how many chars in text are consonants
```

Write a Python program to find the character that appears most frequently in a string. If multiple characters have the same highest frequency, return the first one encountered. Ignore spaces.

* Step - 1 : Observe and Read The Problem
* Step - 2 : Understand the problem
* Step - 3 : whats happening to input to get the desired output
 * Input - string, Output - most frequent character - Single character
* Step - 4 :  Breakdown to simple steps
* Step - 5 : Logic - Pseudocode Intially
    * two variables one is count_char for storing key value pairs of character and thier respective counts and other is counts for storing only counts of characters
    * first loop to traverse through each char and updating char and counts into dictionary count_chars
    * second loop to iterate along the each and every char in a string to add only counts to counts
    * third loop to return most frquent character
    * Conditions to check each and every char for updating counts and checking highest count and returning most frequent character
    * return the most frequent character
* Step - 7 : Test & debug and make changes in code till desired result is returned

```python
def most_frequent_char(text : str):

    count_char = {} # dictionary to store character and counts 

    counts = [] # list to store counts of characters for comparison
     
    for i in text: # Iterating and checking every character in the string and storing each character and its count

        if i in count_char: # Condition checking presence of char in count_char
            
            count_char[i] += 1 # storing achacter and its count
        else:

            count_char[i] = 1 # If not present in count_char the count will be 1

    for char, value in count_char.items(): # Loop to store only counts of each charater 

        counts.append(value) # storing counts 

    for char, value in count_char.items(): # loop to check char which is most repeated 
    
        if value == max(counts): # condition to check if char is the most repeated
            
            return char # returning the first most repeated character
```

```python
def most_repeated_char(text : str):

    counts = []

    for i in text:

        counts.append(text.count(i)) # Appending counts of each character to list

    for char in text:

        if text.count(char) == max(counts): # checking if count of specific character is the highest to identify first most frequent character

            return char # returning first most frequent character
```

```python
def more_than_once(text: str):
    seen = set() # to store characters already seen and using it handle dupliactes in a result list 
    duplicates = [] # chars appeared more than once in a string 
    
    for char in text: # loop to traverse along each and every char 
        
        if char != " " and text.count(char) > 1: # condition to avoid " " amd ahving chars whose count > 1
            
            if char not in seen: # For Avoiding elements which are already appended to list 
                
                seen.add(char)
                
                duplicates.append(char)
                
    return duplicates 
```

### Q16. Find the Most Frequent Word in a Sentence

Write a Python program to find the word that appears most frequently in a sentence. Treat uppercase and lowercase versions of a word as the same word. If multiple words share the highest frequency, return the one that appears first.

Example input:
Python is easy and Python is powerful

Expected output:
python

* Step - 1 : Observe and Read The Problem
* Step - 2 : Understand the problem
* Step - 3 : whats happening to input to get the desired output
 * Input - String, Output - Most Frequent Word in a Sentence - String
* Step - 4 :  Breakdown to simple steps
* Step - 5 : Logic - Pseudocode Intially
    * One varaible to split and store as individual words in a list to iterate over them 
    * Second variable to store counts of each word
    * 1st loop to append counts of words into count_words
    * 2nd loop to return most frequent word using max function and count matching in if condition
    * returning most frequent word
* Step - 7 : Test & debug and make changes in code till desired result is returned

```python
def most_frequent_word(text): 

    list_words = text.lower().split() # Using split and lower methods to handle and store
                                     # each and every word in a list

    count_words = [] # variable that store word counts/ frequency
        
    for i in list_words: # 1st loop to iterate and store counts of words in a list

        count_words.append(list_words.count(i)) # appending count

    for word in list_words: # 2nd loop to check word count and match max count 

        if list_words.count(word) == max(count_words): # condition that checks is the count of word == max count 

            return word # Rettuning most frequent word i.e word with highest frequency
```

### Q17. Find All Words Appearing More Than Once

Write a Python program to find all words that appear more than once in a sentence. Display each repeated word only once, in the order of its first appearance. Treat uppercase and lowercase versions as the same word.

Example input:
data science uses data and science tools

Expected output:
['data', 'science']


* Step - 1 : Observe and Read The Problem
* Step - 2 : Understand the problem
* Step - 3 : whats happening to input to get the desired output
 * Input - String, Output - All Words Appearing More Than Once - List
* Step - 4 :  Breakdown to simple steps
* Step - 5 : Logic - Pseudocode Intially
    * One varaible to split and store as individual words in a list to iterate over them 
    * Second variable to make note of seen words to handle duplicates in result
    * for loop to iterate over all words
    * nested if - outer if for checking is word appearing more than once inner if to handle duplicates
    * If word not in seen it will be added to it and added to result thus handling duplicates
    * returning Words Appearing More Than Once
* Step - 7 : Test & debug and make changes in code till desired result is returned

```python
def more_than_once(text):

    text_lst = text.lower().split() # Using split and lower methods to handle and store
                                     # each and every word in a list

    seen = set() # to keep track words already seen to handle dupliactes 
    
    result = [] # to store words that are appeared more than once

    for word in text_lst: # Loop to iterate over each and every word

        if text_lst.count(word) > 1: # filter words if they are appearing more than once 

            if word not in seen: # handling duplicates with this conmdition 

                seen.add(word)
                
                result.append(word) # appending if its not in result and appeared more than once 
    
    return result # returning words appeared more than once avoiding dupliacte entries in result
```

```python
def word_only_once(text):

    list_words = text.lower().split() # Using split and lower methods to handle and store
                                     # each and every word in a list

    for word in list_words: # loop to iterate over all words

        if list_words.count(word) == 1: # condition to check if word is actually appearing only once

            return word # Returning the first word that appeared only once
```

### Q19. Count Words Whose Length Is Greater Than 4

Write a Python program to count the words containing more than four characters in a sentence.

Example input:
Python makes problem solving interesting

Expected output:
4

* Step - 1 : Observe and Read The Problem
* Step - 2 : Understand the problem
* Step - 3 : whats happening to input to get the desired output
 * Input - String, Output - Count of Words Whose Length Is Greater Than 4 - single Number
* Step - 4 :  Breakdown to simple steps
* Step - 5 : Logic - Pseudocode Intially
    * One varaible to split and store as individual words in a list to iterate over them
    * Second varaiable to store result as a list
    * For loop to iterate over all words
    * If condition to check whether the word length is actually > 4 and append the words which satisfy the condition and return its count
    * Return the count of words whose length is greater than 4
* Step - 7 : Test & debug and make changes in code till desired result is returned

```python
def word_len_gt_four(text):

    list_words = text.split() # Using split to store words as individual words as a list
                                    

    result = [] # to store words whose length > 4
    
    for word in list_words: # loop to iterate over all words 

        if len(word) > 4: # Condition to check if len(word) > 4 or not

            result.append(word) # appending the words with length > 4
            
    return len(result) # Returning count of words whose length > 4
```

```python
# Test Case - 1

word_len_gt_four('Python makes problem solving interesting')
```

```python
# Test Case - 2

word_len_gt_four('Innomatics Research Labs innomatics labs')
```

### Q20. Find the Longest Word That Appears Only Once

Write a Python program to find the longest word that appears exactly once in a sentence. If multiple qualifying words have the same maximum length, return the first one encountered. Treat uppercase and lowercase versions as the same word.

Example input:
Python makes coding useful Python

Expected output:

* Step - 2 : Understand the problem
* Step - 3 : whats happening to input to get the desired output
 * Input - String, Output - Count of Words Whose Length Is Greater Than 4 - single Number

* Step - 5 : Logic - Pseudocode Intially
    * One varaible to split and store as individual words in a list to iterate over them
    * Second varaiable to intiate the largest word and store and return the largest word
    * For loop to iterate over all words
    * Inner If condition to check whether the word length is actually greater than remaining words and update the largest length word
    * Outer IF checks if the word is actually repeating only once in string
    * Return the Longest Word That Appears Only Once
* Step - 7 : Test & debug and make changes in code till desired result is returned

```python
def max_len_word(text):

    list_words = text.split() # Using split to store words as individual words as a list

    largest_word = "" # intiate the largest word, store and return the largest word

    for word in list_words: # Loop to iterate and carry all the words for further steps

        if list_words.count(word) == 1: # IF checks if the word is actually repeating only once in string
        
            if len(word) > len(largest_word): # f condition to check whether the word length is actually greater than remaining words

                largest_word = word # Updating word with highest length
    
    return largest_word # Returning the Longest Word That Appears Only Once
```
